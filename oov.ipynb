```python
import pandas as pd
import json
import re
from collections import Counter

# === Load CSV ===
df = pd.read_csv("all_tokenized.csv")
df["tokens"] = df["text"].apply(lambda x: eval(x))

# === Collect all tokens ===
all_tokens = [token.lower() for tokens in df["tokens"] for token in tokens]
unique_tokens = sorted(set(all_tokens))

print("Total unique tokens:", len(unique_tokens))

# === Load KBBI dictionary ===
with open("dictionary/list_1.0.0_cleaned.txt", "r", encoding="utf-8") as f:
    kbbi_words = set(line.strip().lower() for line in f if line.strip())

# === Load existing normalization dictionary (dict3) ===
with open("dictionary/dict_template3_doneig.json", "r", encoding="utf-8") as f:
    existing_dict = json.load(f)

existing_dict_keys = set(existing_dict.keys())

# === Load extended stopwords ===
with open("stopwords2.txt", "r", encoding="utf-8") as f:
    stopwords = set(line.strip().lower() for line in f if line.strip())

# === Find OOV words ===
oov = [
    w for w in unique_tokens
    if w not in kbbi_words
]

print("Raw OOV count:", len(oov))

# === Remove words already handled ===
filtered_oov = [
    w for w in oov
    if w not in existing_dict_keys
    and w not in stopwords
]

print("Final OOV count (after cross-reference):", len(filtered_oov))

# === Create new slang template ===
slang_template = {word: "" for word in filtered_oov}

with open("dictionary/dict4.json", "w", encoding="utf-8") as f:
    json.dump(slang_template, f, indent=4, ensure_ascii=False)

print("dict4.json created successfully.")
```
