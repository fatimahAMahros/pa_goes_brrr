```python
import pandas as pd

msa = pd.read_csv(
    'wn-msa-all.tab',
    sep='\t',
    header=None,
    names=['synset_id', 'lang', 'goodness', 'lemma'],
    dtype=str
)

msa_indo = msa[msa['lang'].isin(['I', 'B'])].copy()

with open('list_1.0.0_cleaned.txt', 'r', encoding='utf-8') as f:
    kbbi = set(line.strip().lower() for line in f if line.strip())

msa_indo['lemma_lower'] = msa_indo['lemma'].str.lower().str.strip()
msa_filtered = msa_indo[msa_indo['lemma_lower'].isin(kbbi)].copy()

msa_filtered.to_csv('wn_indo_filtered.csv', index=False)

print(f"Total MSA entries:         {len(msa)}")
print(f"After lang filter (I/B):   {len(msa_indo)}")
print(f"After KBBI cross-ref:      {len(msa_filtered)}")
print(f"Unique synsets retained:   {msa_filtered['synset_id'].nunique()}")
print(f"Unique lemmas retained:    {msa_filtered['lemma_lower'].nunique()}")
```

```python
import pandas as pd

msa = pd.read_csv(
    'wn-msa-all.tab',
    sep='\t',
    header=None,
    names=['synset_id', 'lang', 'goodness', 'lemma'],
    dtype=str
)

msa_indo = msa[
    msa['lang'].isin(['I', 'B']) &
    msa['goodness'].isin(['Y', 'O', 'M'])  # change to ['Y', 'O', 'M'] if too small
].copy()

with open('list_1.0.0_cleaned.txt', 'r', encoding='utf-8') as f:
    kbbi = set(line.strip() for line in f if line.strip())

def kbbi_filter(row):
    lemma = row['lemma'].strip()
    # if ' ' in lemma or '-' in lemma:
    #     return True
    return lemma in kbbi

msa_filtered = msa_indo[msa_indo.apply(kbbi_filter, axis=1)].copy()

msa_filtered.to_csv('wn_indo_filtered.csv', index=False)

print(f"Total MSA entries:         {len(msa)}")
print(f"After lang+goodness filter:{len(msa_indo)}")
print(f"After KBBI cross-ref:      {len(msa_filtered)}")
print(f"Unique synsets retained:   {msa_filtered['synset_id'].nunique()}")
print(f"Unique lemmas retained:    {msa_filtered['lemma'].nunique()}")

print("\nGoodness distribution in filtered set:")
print(msa_filtered['goodness'].value_counts())

domain_words = ['mati', 'nyala', 'padam', 'macet', 'bocor', 'rusak']
print("\nDomain word coverage check:")
for word in domain_words:
    in_msa  = word in msa_indo['lemma'].str.strip().values
    in_filt = word in msa_filtered['lemma'].str.strip().values
    print(f"  {word:10s} | in MSA(I/B)+quality: {str(in_msa):5} | survives KBBI filter: {in_filt}")
```
