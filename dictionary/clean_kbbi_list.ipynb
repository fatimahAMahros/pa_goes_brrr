```python
import re

cleaned = []

with open("list_1.0.0.txt", encoding="utf-8") as f:
    for line in f:
        w = line.strip()

        if " " in w:
            continue

        if re.search(r"[.,()/:;!?]", w):
            continue

        if len(w) > 20:
            continue

        if re.search(r"\d", w):
            continue

        if not w:
            continue

        cleaned.append(w)

cleaned = sorted(set(cleaned))

with open("list_1.0.0_cleaned.txt", "w", encoding="utf-8") as f:
    for w in cleaned:
        f.write(w + "\n")
```
